```python
import pandas as pd
```

```python
logs = pd.read_csv("../data/logs.csv")

logs
```

```python
features = [
    "login_count",
    "failed_logins",
    "traffic_mb",
    "session_duration"
]

X = logs[features]

X
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled[:5]
```

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(
    eps=0.5,
    min_samples=5
)
```

```python
labels = dbscan.fit_predict(X_scaled)

print(labels)
```

```python
logs["cluster"] = labels

logs
```

```python
logs["is_anomaly"] = logs["cluster"] == -1

logs
```

```python
logs[logs["is_anomaly"]]
```

```python
print("Number of anomalies:", logs["is_anomaly"].sum())
```
